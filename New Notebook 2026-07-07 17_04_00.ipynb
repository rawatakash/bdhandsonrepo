```python
%sql
select ProductCategory,sum(SalesAmount) as Total_Sales,sum(SalesQuantity) as total_quantity,sum(SalesAmount)/sum(SalesQuantity) as avgsellingprice from parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/Sales.parquet` s
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/Calendar.parquet` c on s.DateKey=c.Datekey
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/Channel.parquet` ch on s.channelKey=ch.channel
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/Stores.parquet` st on s.Storekey=st.Storekey
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/Geography.parquet` geo on st.GeographyKey=geo.GeographyKey
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/Product.parquet` pro on s.Productkey=pro.ProductKey
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/ProductSubcategory.parquet` prosubat on pro.ProductSubCategoryKey=prosubat.ProductSubCategoryKey
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/ProductCategory.parquet` procat on prosubat.ProductCategoryKey=procat.ProductCategoryKey
left join parquet.`abfss://bronze@storageaccbd.dfs.core.windows.net/Promotion.parquet` promotion on s.PromotionKey=promotion.PromotionKey
group by ProductCategory
```
